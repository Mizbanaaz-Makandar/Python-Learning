**Excercise 1: Prime Numbers**

```python
def is_prime(n):
  if n <= 1:
    return False

  for i in range(2, int(n ** 0.5) + 1):
    if n % i == 0:
      return False
  return True
```

```python
num = int(input("Enter a number : "))
if is_prime(num):
  print(f"{num} is a prime number.")
else:
  print(f"{num} is not a prime number.")
```

```python
# Alternative solution

number = int(input("Enter a number: "))

if number <= 1:
  print("Not a prime number")
else:
  is_prime = True

  for i in range(2, number):
    if number % i == 0:
      is_prime = False
      break

  if is_prime:
    print("Prime number")
  else:
    print("Not a prime number")
```

**Excercise 2 : Product of Random Numbers**

```python
import random

number1 = random.randint(1,10)
number2 = random.randint(1,10)

correct_answer = number1 * number2

user_answer = int(input(f"What is {number1} X {number2} ?"))

if user_answer == correct_answer:
  print("Correct answer!")
else:
  print("Incorrect answer.")
  print("The correct answer is",correct_answer)
```

**Excercise 3 : Square of Even/ Odd numbers**

```python
print("Square of even numbers from 100 to 200: ")

for num in range(100,201):
  if num % 2 == 0:  # even check
    print(f"{num}2 = {num ** 2}")
```

**Exercise 4: Hospital Billing System**

```python
# Given constants
ROOM_CHARGE_PER_DAY = 2000
DOCTOR_FEE = 1500
LAB_TEST_CHARGE = 300

# Patient details
days_stayed = 4
num_lab_tests = 3
medicine_charges = 2400

# Calculate the bill
room_total = days_stayed * ROOM_CHARGE_PER_DAY
lab_total = num_lab_tests * LAB_TEST_CHARGE

total_before_discount = room_total + DOCTOR_FEE + lab_total + medicine_charges

# Discount 10% if total exceed 10,000
if total_before_discount > 10000:
  discount = total_before_discount * 0.10
else:
  discount = 0

final_bill = total_before_discount - discount

# Bill
print("=" * 40)
print("        HOSPITAL BILL SUMMARY")
print("=" * 40)
print(f"Room charges ({days_stayed} days × ₹{ROOM_CHARGE_PER_DAY}) : ₹{room_total}")
print(f"Doctor consultation fee          : ₹{DOCTOR_FEE}")
print(f"Lab tests ({num_lab_tests} × ₹{LAB_TEST_CHARGE})           : ₹{lab_total}")
print(f"Medicine charges                 : ₹{medicine_charges}")
print("-" * 40)
print(f"Total before discount             : ₹{total_before_discount}")
print(f"Discount (10%)                    : ₹{discount:.2f}")
print("-" * 40)
print(f"FINAL BILL                        : ₹{final_bill:.2f}")
print("=" * 40)
```
