```python
# Conditional Statement: Program to find out which number is greatest among two values
a=1500
b=1450
if a>b:
  print("a is greater")
  print("In True block")
  c=a+b
  print(c)
else:
  print('b is greater')
  print("In False block")
  c=a*b
  print(c)

# By taking input from user:

num = eval(input("Enter a number: "))

if num > 0:
    print("Positive number")

elif num < 0:
    print("Negative number")

else:
    print("Neutral")
```

```python
# Print all even and odd numbers between given range using For Loop
print('Even Numbers:')
for i in range(1, 41):
  if i % 2 == 0:
    print(i, end=" ")

print(" ") # new line
print('Odd Numbers:')
for i in range(1,41):
  if i%2!=0: # != not equal to
    print(i, end=" ")
```
