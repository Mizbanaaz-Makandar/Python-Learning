```python
def greet(name):
    message = f"Hello,{name}! Welcome to Data Science."
    return message

print(greet("Alice"))
print(greet("Bob"))
```

```python
# A data science function (calculate discount)

def apply_discount(price,discount_percentage):
    discount_amount = price * (discount_percentage/100)
    final_price = price - discount_amount
    return final_price

# Calling the function for different products:
laptop_price = apply_discount(1000,10)  # 10% off
phone_price = apply_discount(500,15)   # 15% off

print(f"Laptop final price: ${laptop_price}")
print(f"Phone final price: ${phone_price}")
```

```python
def calculate_average(numbers):
    total = sum(numbers)
    count = len(numbers)
    average = total / count
    return average
```

```python
scores = [85,90,78,92,88]
avg = calculate_average(scores)
print(f"The average score is : {avg}")
```

```python
# Tax calculator function 8%
def tax_calculator(amount, tax_rate = 0.08): 
    total = amount + (amount * tax_rate)
    return total
```

```python
print(tax_calculator(100))
print(tax_calculator(100 , 0.05))
print(tax_calculator(250, 0.10))
```

```python
# Alternative
def tax_only(amount, tax_rate = 0.08):
    return amount * tax_rate

print(tax_only(100))
```

```python
# Solution: Function that returns a new list with scores > 70
def filter_high_scores(scores):
    high_scores = []  # Create an empty list to store results
    for score in scores:
        if score > 70:
            high_scores.append(score)  # Add to the new list
    return high_scores

# Testing the function:
test_scores = [45, 88, 62, 95, 70, 81]
result = filter_high_scores(test_scores)
print(result)  # Output: [88, 95, 81]

# More advanced solution using list comprehension (1 line!)
def filter_high_scores_v2(scores):
    return [score for score in scores if score > 70]

# Testing both versions (they give the same result):
print(filter_high_scores_v2([30, 75, 60, 90, 55]))  # Output: [75, 90]
```
