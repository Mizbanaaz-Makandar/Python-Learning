```python
fruits = {"apple","Kiwi"}
fruits.add("Melon")
print(fruits)

# To update: to add multiple elements

fruits.update(["mango","orange"])
print(fruits)

# pop method:
item = fruits.pop()
print(item)

# copy() :
s1 = {1,2,3}
s2 = s1.copy()
s2

# union: addition on both the sets but it will consider comman element
A = {1,2,3}
B ={3,4,5}

print(A.union(B))
print(B | A)

# Intersection : results comman elements

print(A.intersection(B))
print(A & B)

# difference() : It will return elements present in the first set but not in the second

print(A.difference(B))
print(A - B)

# Symmetric difference: It will return elements that are in either set but not in both

print(A.symmetric_difference(B))
print(A ^ B)

# issubset() : Checks whether one set is a subset of nother

a = {1,2}
b = {1,2,3,4}
print(a.issubset(b))

# issuperset: checks whether one set contains another setA
x = {1,2,3,4}
y = {1,2}
print(x.issuperset(y))

# membership operator(in): 

fruits = {"apple","mango","kiwi"}
print("apple" in fruits)
print("grapes" in fruits)
```

```python
# Nested conditional statements:

age = 16
has_id = True

if age >= 18:
    if has_id:
        print("You can vote!")
    else:
        print("Id is required")
else:
    print("Underage")   
```
